```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]
```

```python
# Etape 1 - une seule d'abord
import requests

url = "https://api-adresse.data.gouv.fr/search/"
r = requests.get(url=url, params={"q": adresses[0], "limit": 1}, timeout=10)

print(f"Statut : {r.status_code}")
print(f"Label : {r.json()["features"][0]["properties"]["label"]}")
print(f"Score : {r.json()["features"][0]["properties"]["score"]}")
```

```python
liste_adresses = []

for adresse in adresses:
    url = "https://api-adresse.data.gouv.fr/search/"
    r = requests.get(url=url, params={"q": adresse, "limit": 1}, timeout=10)
    donnees = r.json()

    if donnees.get("features"):
        liste_adresses.append({
            "adresse": adresse,
            "label": donnees["features"][0]["properties"]["label"],
            "score": donnees["features"][0]["properties"]["score"],
            "postcode": donnees["features"][0]["properties"]["postcode"],
            "city": donnees["features"][0]["properties"]["city"],
            })
    else:
        liste_adresses.append({
            "adresse": adresse,
            "label": None,
            "score": None,
            "postcode": None,
            "city": None,
            })

print(liste_adresses)
```

```python
# Etape 2 - les cinq (autre solution)
liste_adresses = []

for adresse in adresses:
    url = "https://api-adresse.data.gouv.fr/search/"
    r = requests.get(url=url, params={"q": adresse, "limit": 1}, timeout=10)
    donnees = r.json()

    if donnees.get("features"):
        proprietes = donnees["features"][0]["properties"]

        liste_adresses.append({
            "adresse": adresse,
            "label": proprietes.get("label"),
            "score": proprietes.get("score"),
            "postcode": proprietes.get("postcode"),
            "city": proprietes.get("city"),
            })
    else:
        liste_adresses.append({
            "adresse": adresse,
            "label": None,
            "score": None,
            "postcode": None,
            "city": None,
            })

print(liste_adresses)
```

```python
# Etape 3 - en DataFrame
import pandas as pd

df_adresses = pd.DataFrame(liste_adresses)

df_adresses
```

```python
# Etape 4 - lire le résultat
correct = len(df_adresses) - df_adresses["label"].isna().sum()

print(f"Nombre d'adresses trouvées : {correct}")
print()
print(f"Adresse avec le score le plus bas : {df_adresses["adresse"][df_adresses["score"] == df_adresses["score"].min()].to_string(index=False)}")
print("C'est cette adresse qui a le plus faible score car le label renvoyé ne correspond pas exactement à l'adresse recherchée.")
print("Il y a beaucoup de différences entre les deux et une vérification humaine est nécessaire")
print()
print("Le label ne correspond pas toujours à ce que l'on demande. L'API renvoie ce qu'elle trouve de plus proche.")
```
